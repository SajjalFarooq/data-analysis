# Data Analysis with Pandas — Complete Guide

This notebook covers pandas completely, step by step, for beginners. Every new code type is explained with a mini definition in the text above it, and with proper comments inside the code, so every beginner can understand why each piece of code is used.

**Requirements:** Python 3, pandas, matplotlib and numpy. These are the only external packages used.

**How to run:** Open this notebook in Jupyter, VS Code, or Google Colab and run all cells. This notebook was executed before upload, so every output below is a real output.

## 1. Series — One Column of Data

**Mini definition:** A Series is a single column of data with an index, like one list with labels. A DataFrame is made of many Series put together.

```python
# pandas is imported as pd by convention, so pd.Series is short to write
import pandas as pd

# Series() turns a list into a single column, with an automatic index 0, 1, 2...
marks_series = pd.Series([86, 69, 95, 78, 88])
print(marks_series)
print()
# mean() gives the average of the Series, so you can check the centre of the data
print("Mean marks:", marks_series.mean())
```

## 2. DataFrame from a Dictionary

**Mini definition:** A DataFrame is a table with rows and columns, like a spreadsheet. Each column has a name, and each row is one record, such as one student.

```python
# A dictionary is used here: keys become column names, lists become column values
data = {
    "Name": ["Sajjal", "Ali", "Ayesha", "Bilal", "Fatima"],
    "Field": ["Computer Science", "Mathematics", "Computer Science", "Mathematics", "Computer Science"],
    "Marks": [86, 69, 95, 78, 88],
}
# DataFrame() turns the dictionary into a table object called df
df = pd.DataFrame(data)
print(df)
```

## 3. Reading a CSV File

**Mini definition:** A CSV file is a plain text file where values are separated by commas, one row per line. `read_csv()` loads a CSV file into a DataFrame, so you can analyse data saved outside Python.

```python
# read_csv() reads students.csv from the same folder into a DataFrame called df_csv
# students.csv is included in this repository, so this cell works when you run it
df_csv = pd.read_csv("students.csv")
print(df_csv)
```

## 4. Viewing Data: head, tail, shape, columns, info

**Mini definition:** `head()` shows the first few rows, `tail()` shows the last few rows, `shape` tells you (rows, columns), `columns` lists the column names, and `info()` shows each column's data type and how many values are missing.

```python
# head(3) shows only the first 3 rows, so the output stays short
print(df_csv.head(3))
print()
# tail(2) shows only the last 2 rows
print(df_csv.tail(2))
print()
# shape is an attribute, not a method, so no parentheses are used here
print("Shape (rows, columns):", df_csv.shape)
print("Columns:", list(df_csv.columns))
```

```python
# info() prints a summary: column names, non-null counts and data types
# It is used to check quickly if any column has missing values
df_csv.info()
```

## 5. Descriptive Statistics

**Mini definition:** `describe()` gives a summary of the numeric columns: count, mean (average), standard deviation, minimum, maximum, and quartiles. It is used to understand the data quickly, before any deeper analysis.

```python
# describe() works on numeric columns, so it summarises Marks and Age here
print(df_csv.describe())
```

## 6. Selecting Columns and Rows: loc and iloc

**Mini definition:** Selecting means picking part of a table. `loc` selects by label, such as a column name or a row label. `iloc` selects by position, such as row number 0 or column number 1, starting from zero.

```python
# df_csv["Marks"] selects one column by its name, and returns a Series
print(df_csv["Marks"])
print()
# loc[0, "Name"] selects the row with label 0 and the column called Name
print("loc[0, Name]:", df_csv.loc[0, "Name"])
print()
# iloc[0, 2] selects the row at position 0 and the column at position 2 (Marks)
print("iloc[0, 2]:", df_csv.iloc[0, 2])
```

## 7. Filtering Rows with a Condition

**Mini definition:** Filtering means keeping only the rows that meet a condition, and dropping the rest. A condition like `Marks > 80` creates a True/False value for each row, and pandas keeps only the True rows.

```python
# df_csv["Marks"] > 80 creates True for high marks and False for the rest
# Putting that inside df_csv[...] keeps only the True rows
high_marks = df_csv[df_csv["Marks"] > 80]
print(high_marks)
```

## 8. Sorting Data

**Mini definition:** `sort_values()` sorts rows by a column, from smallest to largest, or largest to smallest if ascending is False. Sorting is used to find the top or bottom records easily.

```python
# sort_values() sorts by Marks, largest first, so the top student is first
sorted_df = df_csv.sort_values(by="Marks", ascending=False)
print(sorted_df)
```

## 9. Adding, Renaming and Dropping Columns

**Mini definition:** A new column can be created from existing columns, for example Pass/Fail from Marks. `rename()` changes a column name. `drop()` removes a column or row. All of these are used to clean and shape data for analysis.

```python
# A copy is made first, so the original df_csv is not changed by this example
df2 = df_csv.copy()
# A new column is created: Pass is True if Marks are 60 or more, else False
# This is a boolean column, made by comparing Marks with 60
df2["Pass"] = df2["Marks"] >= 60
# rename() changes the column name Marks to Score, for this example only
# inplace=False is the default, so a new DataFrame is returned, not changed in place
df2 = df2.rename(columns={"Marks": "Score"})
print(df2)
print()
# drop() removes the Age column here, axis=1 means column, not row
print(df2.drop(columns=["Age"]))
```

## 10. Handling Missing Values: isna, fillna and dropna

**Mini definition:** A missing value means a cell has no data, shown as NaN (Not a Number) in pandas. `isna()` finds missing values, `fillna()` replaces them with a chosen value, and `dropna()` removes rows that have missing values. You choose one based on whether the row is still useful without that value.

```python
# A copy is made, then one cell is set to NaN on purpose, for this example
import numpy as np
df_missing = df_csv.copy()
df_missing.loc[1, "Marks"] = np.nan
print("With a missing value:")
print(df_missing)
print()
# isna() returns True where a value is missing, and sum() counts True values per column
print("Missing values per column:")
print(df_missing.isna().sum())
print()
# fillna() replaces NaN with the mean of Marks, so calculations do not fail
mean_marks = df_missing["Marks"].mean()
df_filled = df_missing.fillna({"Marks": mean_marks})
print("After fillna with the mean:")
print(df_filled)
print()
# dropna() removes any row that has a missing value, instead of filling it
print("After dropna:")
print(df_missing.dropna())
```

## 11. Removing Duplicates

**Mini definition:** A duplicate row is an exact copy of another row. `duplicated()` finds them, and `drop_duplicates()` removes them, so each record is counted only once.

```python
# A copy is made, then the first row is added again, to create a duplicate
df_dup = pd.concat([df_csv, df_csv.iloc[[0]]], ignore_index=True)
print("With a duplicate row:")
print(df_dup)
print()
# drop_duplicates() keeps only the first copy of each row and removes the rest
print("After drop_duplicates:")
print(df_dup.drop_duplicates())
```

## 12. Grouping and Aggregation

**Mini definition:** `groupby()` groups rows that share the same value in a column, for example all Computer Science students together. After grouping, an aggregate like `mean()`, `count()` or `max()` can be calculated for each group separately. `agg()` lets you calculate several aggregates at once.

```python
# groupby("Field") groups by field, then mean() gives average Marks per group
print(df_csv.groupby("Field")["Marks"].mean())
print()
# agg() calculates mean, max and count together, in one step, per group
print(df_csv.groupby("Field")["Marks"].agg(["mean", "max", "count"]))
```

## 13. Value Counts

**Mini definition:** `value_counts()` counts how many times each unique value appears in a column. It is used to see the distribution of a category, for example how many students are in each field.

```python
# value_counts() counts each Field value: how many Computer Science, how many Mathematics
print(df_csv["Field"].value_counts())
```

## 14. Apply: Running a Function on Every Value

**Mini definition:** `apply()` runs a function on every value in a column, one by one. It is used when a simple calculation is not enough, and you need custom logic, such as turning marks into grades.

```python
# A function is defined first: it turns one mark into one grade letter
def mark_to_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "F"

# apply() runs mark_to_grade on every value in the Marks column, and stores the result
# in a new column called Grade, so the original Marks column is not changed
df_csv["Grade"] = df_csv["Marks"].apply(mark_to_grade)
print(df_csv)
```

## 15. Merging Two DataFrames

**Mini definition:** Merging joins two tables on a common column, like joining student marks with student cities on Name. `merge()` is used when related data is stored in two separate tables and you need one complete table.

```python
# A second table is created: each student's city, with Name as the common column
cities = pd.DataFrame({
    "Name": ["Sajjal", "Ali", "Ayesha", "Bilal", "Fatima", "Omar"],
    "City": ["Gujrat", "Lahore", "Gujrat", "Islamabad", "Lahore", "Karachi"],
})
# merge() joins df_csv and cities on Name, so each student now also has a City
merged_df = pd.merge(df_csv, cities, on="Name")
print(merged_df)
```

## 16. Pivot Table

**Mini definition:** A pivot table summarises data in a grid: rows are one category, columns are another category, and each cell shows an aggregate, such as the mean. It is used to compare two categories at once, for example Field vs Grade.

```python
# pivot_table() makes Field the rows, Grade the columns, and Marks the values
# aggfunc="mean" means each cell shows the average Marks for that Field and Grade
pivot = merged_df.pivot_table(index="Field", columns="Grade", values="Marks", aggfunc="mean")
print(pivot)
```

## 17. Creating a Bar Chart

**Mini definition:** A bar chart shows one bar for each group, and the height of each bar shows its value. It is used to compare groups at a glance. Here, each bar shows the average marks for one field of study.

```python
# matplotlib is imported as plt by convention, for plotting
import matplotlib
matplotlib.use("Agg")  # Agg saves the chart to a file, without opening a window
import matplotlib.pyplot as plt

# average_by_field is calculated again here, so this cell works on its own
average_by_field = df_csv.groupby("Field")["Marks"].mean()
# plot() with kind="bar" draws a bar chart from the grouped averages
average_by_field.plot(kind="bar", color=["#4CAF50", "#2196F3"])
# plt.title and labels explain what the chart shows, so it is readable alone
plt.title("Average Marks by Field")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
# savefig() saves the chart as a file in the same folder as this notebook
plt.savefig("average-marks-by-field.png")
plt.close()
print("Chart saved as average-marks-by-field.png")
print(average_by_field)
```
